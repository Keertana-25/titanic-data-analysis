```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("train.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df=df.drop(columns=['Cabin'])
```

```python
df.isnull().sum()
```

```python
df['Embarked']=df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.groupby('Sex')['Survived'].mean()
```

```python
df.groupby('Pclass')['Survived'].mean()
```

```python
bins = [0, 12, 20, 40, 60, 80]
labels = ['Child', 'Teen', 'Adult', 'Middle-Age', 'Senior']

df['AgeGroup'] = pd.cut(df['Age'], bins=bins, labels=labels)
df.groupby('AgeGroup')['Survived'].mean()
```

```python
df.groupby('AgeGroup', observed=True)['Survived'].mean()
```

### Final Insights:
- Females had a significantly higher survival rate than males.
- Passenger class strongly affected survival, with first-class passengers surviving more.
- Children had better survival chances compared to adults and seniors.
- Senior passengers had the lowest survival rate.

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(x='Sex', y='Survived', data=df)
plt.title("Survival by Gender")
plt.show()
```

```python
plt.hist(df['Age'], bins=20)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution")
plt.show()
```

```python
sns.barplot(x='Pclass', y='Survived', data=df)
plt.title("Survival by Passenger Class")
plt.show()
```

### Insights:
- Females had a significantly higher survival rate than males.
- Passengers in first class survived more than those in second and third class.
- Children had better survival chances compared to adults.
